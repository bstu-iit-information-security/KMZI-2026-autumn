```python
print("="*50)
print("МОДУЛЬ 2: ЛОГИЧЕСКАЯ ФУНКЦИЯ (Задание 2, Вариант 6)")
print("="*50)

def build_classic_f(a, b, c, d):
    maj = (a & b) | (a & c) | (b & c)
    return int(maj & (not d))

# 2.1 Базовая схема (7 кубитов)
def build_basic_circuit():
    qr = QuantumRegister(7, 'q'); cr = ClassicalRegister(7, 'c')
    qc = QuantumCircuit(qr, cr)
    qc.ccx(0, 1, 4); qc.ccx(0, 2, 4); qc.ccx(1, 2, 4) # MAJ
    qc.cx(3, 5); qc.x(5) # NOT D
    qc.ccx(4, 5, 6) # F
    qc.x(5); qc.cx(3, 5) # Uncompute NOT D
    qc.ccx(1, 2, 4); qc.ccx(0, 2, 4); qc.ccx(0, 1, 4) # Uncompute MAJ
    qc.measure(qr, cr)
    return qc

# 2.2 Оптимизированная схема (6 кубитов)
def build_optimized_circuit():
    qr = QuantumRegister(6, 'q'); cr = ClassicalRegister(6, 'c')
    qc = QuantumCircuit(qr, cr)
    qc.ccx(0, 1, 4); qc.ccx(0, 2, 4); qc.ccx(1, 2, 4)
    qc.x(3) # NOT D
    qc.ccx(4, 3, 5)
    qc.x(3) # Возврат D
    qc.ccx(1, 2, 4); qc.ccx(0, 2, 4); qc.ccx(0, 1, 4)
    qc.measure(qr, cr)
    return qc

print("Базовая схема (7 кубитов):")
display(build_basic_circuit().draw('mpl'))
print("Оптимизированная схема (6 кубитов):")
display(build_optimized_circuit().draw('mpl'))

# 2.3 Тестирование
def test_circuit(n_qubits, out_idx, ancilla_indices, name):
    print(f"\n--- Тестирование: {name} ---")
    print(f"{'A':<3} {'B':<3} {'C':<3} {'D':<3} | {'F_class':<8} | {'F_quant':<8} | {'Ancilla':<10} | Status")
    all_ok = True
    for i in range(16):
        a=(i>>0)&1; b=(i>>1)&1; c=(i>>2)&1; d=(i>>3)&1
        f_class = build_classic_f(a,b,c,d)

        qc_run = QuantumCircuit(n_qubits, n_qubits)
        if a: qc_run.x(0)
        if b: qc_run.x(1)
        if c: qc_run.x(2)
        if d: qc_run.x(3)

        qc_base = build_basic_circuit() if n_qubits==7 else build_optimized_circuit()
        qc_run.compose(qc_base.remove_final_measurements(inplace=False), inplace=True)
        qc_run.measure(range(n_qubits), range(n_qubits))

        counts = sim.run(transpile(qc_run, sim), shots=100).result().get_counts()
        most_freq = max(counts, key=counts.get)
        f_quant = int(most_freq[0])
        ancillas = [int(most_freq[-(idx+1)]) for idx in ancilla_indices]

        status = "OK" if (f_quant == f_class and all(v==0 for v in ancillas)) else "FAIL"
        if status == "FAIL": all_ok = False
        print(f"{a:<3} {b:<3} {c:<3} {d:<3} | {f_class:<8} | {f_quant:<8} | {ancillas} | {status}")
    print(f"Итог: {'Все тесты пройдены!' if all_ok else 'Есть ошибки!'}")

test_circuit(7, 0, [2,1], "Базовая (2 ancilla)")
test_circuit(6, 0, [1], "Оптимизированная (1 ancilla)")
```

```python
print("="*50)
print("МОДУЛЬ 3: СИМУЛЯЦИЯ С ШУМОМ (Задание 3)")
print("="*50)

def build_classic_f(a, b, c, d):
    maj = (a & b) | (a & c) | (b & c)
    return int(maj & (not d))

qc_opt = QuantumCircuit(6)
qc_opt.ccx(0, 1, 4); qc_opt.ccx(0, 2, 4); qc_opt.ccx(1, 2, 4)
qc_opt.x(3)
qc_opt.ccx(4, 3, 5)
qc_opt.x(3)
qc_opt.ccx(1, 2, 4); qc_opt.ccx(0, 2, 4); qc_opt.ccx(0, 1, 4)

qc_compiled = transpile(qc_opt, sim, basis_gates=['u3', 'cx'])
print(f"После транспиляции: {qc_compiled.count_ops().get('cx', 0)} гейтов CX.\n")

p2_vals = [0, 0.001, 0.005, 0.01, 0.02]
accuracies = []
for p2 in p2_vals:
    noise_model = NoiseModel()
    noise_model.add_all_qubit_quantum_error(depolarizing_error(0.001, 1), ['u3'])
    if p2 > 0:
        noise_model.add_all_qubit_quantum_error(depolarizing_error(p2, 2), ['cx'])

    correct = 0
    for i in range(16):
        a=(i>>0)&1; b=(i>>1)&1; c=(i>>2)&1; d=(i>>3)&1
        f_class = build_classic_f(a,b,c,d)

        qc_run = QuantumCircuit(6, 6)
        if a: qc_run.x(0)
        if b: qc_run.x(1)
        if c: qc_run.x(2)
        if d: qc_run.x(3)
        qc_run.compose(qc_compiled, inplace=True)
        qc_run.measure(range(6), range(6))

        counts = sim.run(qc_run, noise_model=noise_model, shots=1024).result().get_counts()
        if int(max(counts, key=counts.get)[0]) == f_class:
            correct += 1
    acc = correct / 16
    accuracies.append(acc)
    print(f"p2 = {p2:.3f}: точность = {acc:.1%}")

plt.figure(figsize=(8,5))
plt.plot(p2_vals, accuracies, 'o-', color='blue', linewidth=2)
plt.title('Влияние шума на точность (Вариант 6)')
plt.xlabel('Вероятность p2')
plt.ylabel('Доля правильных ответов')
plt.grid(True); plt.ylim(0, 1.05); plt.show()
```

```python
print("="*50)
print("МОДУЛЬ 5: ИТЕРАЦИИ ГРОВЕРА (Задание 5)")
print("="*50)

def build_oracle(n, marked):
    qc = QuantumCircuit(n)
    for i, bit in enumerate(reversed(marked)):
        if bit == '0': qc.x(i)
    qc.h(n-1); qc.mcx(list(range(n-1)), n-1); qc.h(n-1)
    for i, bit in enumerate(reversed(marked)):
        if bit == '0': qc.x(i)
    return qc

def build_diffuser(n):
    qc = QuantumCircuit(n)
    qc.h(range(n)); qc.x(range(n))
    qc.h(n-1); qc.mcx(list(range(n-1)), n-1); qc.h(n-1)
    qc.x(range(n)); qc.h(range(n))
    return qc

n = 7; marked = '1010011'; N = 2**n
theta = np.arcsin(1/np.sqrt(N))
max_k = 15

P_exact = []
for k in range(max_k+1):
    qc = QuantumCircuit(n)
    qc.h(range(n))
    for _ in range(k):
        qc.compose(build_oracle(n, marked), inplace=True)
        qc.compose(build_diffuser(n), inplace=True)
    sv = Statevector(qc)
    P_exact.append(abs(sv.data[int(marked, 2)])**2)

plt.figure(figsize=(10,5))
plt.plot(range(max_k+1), P_exact, 'o-', label='Точная (Statevector)')
plt.plot(range(max_k+1), [np.sin((2*k+1)*theta)**2 for k in range(max_k+1)], '--', label='Теория sin^2')
plt.xlabel('k'); plt.ylabel('P(k)'); plt.title(f'Гровер для |{marked}>')
plt.legend(); plt.grid(True); plt.show()

k_opt = int(np.floor(np.pi/4 * np.sqrt(N)))
print(f"Теоретически оптимальное k = {k_opt}")
print(f"Экспериментальный максимум при k = {np.argmax(P_exact)}")
```

```python
print("="*50)
print("МОДУЛЬ 6: НЕСКОЛЬКО РЕШЕНИЙ (Задание 6)")
print("="*50)

def build_oracle(n, marked):
    qc = QuantumCircuit(n)
    for i, bit in enumerate(reversed(marked)):
        if bit == '0': qc.x(i)
    qc.h(n-1); qc.mcx(list(range(n-1)), n-1); qc.h(n-1)
    for i, bit in enumerate(reversed(marked)):
        if bit == '0': qc.x(i)
    return qc

def build_diffuser(n):
    qc = QuantumCircuit(n)
    qc.h(range(n)); qc.x(range(n))
    qc.h(n-1); qc.mcx(list(range(n-1)), n-1); qc.h(n-1)
    qc.x(range(n)); qc.h(range(n))
    return qc

n = 6; N = 2**n
M_values = [1, 2, 3, 4, 8]
np.random.seed(42)
all_states = [format(i, f'0{n}b') for i in range(N)]

plt.figure(figsize=(10,6))
for M in M_values:
    marked_states = np.random.choice(all_states, size=M, replace=False)

    P_k = []
    for k in range(16):
        qc = QuantumCircuit(n)
        qc.h(range(n))
        for _ in range(k):
            for state in marked_states:
                qc.compose(build_oracle(n, state), inplace=True)
            qc.compose(build_diffuser(n), inplace=True)
        sv = Statevector(qc)
        prob = sum(abs(sv.data[int(state, 2)])**2 for state in marked_states)
        P_k.append(prob)

    first_max = next(i for i in range(1, len(P_k)-1) if P_k[i] > P_k[i-1] and P_k[i] > P_k[i+1])
    print(f"M={M}: k_теор={int(np.floor(np.pi/4*np.sqrt(N/M)))}, 1-й max при k={first_max} (P={P_k[first_max]:.3f})")
    plt.plot(range(16), P_k, 'o-', label=f'M={M}')

plt.xlabel('k'); plt.ylabel('P'); plt.title('Гровер с M решениями (N=64)')
plt.legend(); plt.grid(True); plt.show()
```

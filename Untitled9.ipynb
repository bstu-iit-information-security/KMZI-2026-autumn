```python
# ============================================================
# ЗАДАНИЕ 1: АЛГОРИТМ ДОЙЧА (ИСПРАВЛЕННЫЙ)
# ============================================================
import numpy as np
import pandas as pd
from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector, partial_trace
from qiskit_aer import AerSimulator

np.random.seed(42)
sim = AerSimulator()

# ---------- Оракулы ----------
def twobit_function_1(qc, q0, q1):  # f(0)=0, f(1)=1  balanced
    qc.cx(q0, q1)
def twobit_function_2(qc, q0, q1):  # f=0  constant
    pass
def twobit_function_3(qc, q0, q1):  # f=1  constant
    qc.x(q1)
def twobit_function_4(qc, q0, q1):  # f(0)=1, f(1)=0  balanced
    qc.x(q1); qc.cx(q0, q1)

ORACLES = {1: twobit_function_1, 2: twobit_function_2,
           3: twobit_function_3, 4: twobit_function_4}
TRUTH = {1:"balanced", 2:"constant", 3:"constant", 4:"balanced"}

def make_oracle(idx):
    qc = QuantumCircuit(2)
    ORACLES[idx](qc, 0, 1)
    g = qc.to_gate(); g.name = f"U_{idx}"
    return g

def deutsch_circuit(gate, measure=True, prep_target=True, h_control=True):
    qc = QuantumCircuit(2, 1 if measure else 0)
    qc.h(0)
    if prep_target: qc.x(1)
    qc.h(1)
    qc.append(gate, [0, 1])
    if h_control: qc.h(0)
    if measure: qc.measure(0, 0)
    return qc

# ---------- 1.1 ----------
print("=== 1.1 ===")
idx = 1
g = make_oracle(idx)
counts = sim.run(transpile(deutsch_circuit(g), sim), shots=1024).result().get_counts()
bit = max(counts, key=counts.get)
ans = "constant" if bit == "0" else "balanced"
print("counts:", counts, "->", ans)
assert ans == TRUTH[idx]
print("'1' на control => фаза (-1) на |+> => |-> => сбалансированная функция.")

# ---------- 1.2 ----------
print("\n=== 1.2 ===")
qc = QuantumCircuit(2)
qc.h(0); qc.x(1); qc.h(1)
qc.append(g, [0, 1])
rho = partial_trace(Statevector(qc), [1]).data
print("rho(qubit0) =\n", np.round(rho.real, 3))
if TRUTH[idx] == "constant":
    assert np.allclose(rho, [[.5,.5],[.5,.5]], atol=1e-6); print("|+> OK")
else:
    assert np.allclose(rho, [[.5,-.5],[-.5,.5]], atol=1e-6); print("|-> OK")
print("Оракул не меняет |0>,|1> в выч. базисе, но добавляет относительную фазу,")
print("которая превращает |+> в |-> (глобальная фаза становится относительной).")

# ---------- 1.3 ----------
print("\n=== 1.3: без X на target ===")
rows = []
for i in [1,2,3,4]:
    c = sim.run(transpile(deutsch_circuit(make_oracle(i), prep_target=False), sim),
                shots=1024).result().get_counts()
    b = max(c, key=c.get)
    a_ = "constant" if b == "0" else "balanced"
    rows.append((i, b, a_, TRUTH[i], a_ == TRUTH[i]))
print(pd.DataFrame(rows, columns=["oracle","bit","answer","truth","ok"]).to_string(index=False))
print("Пояснение: без |1> на target оракул X на target (функции 3,4) работает иначе,")
print("фаза не кодируется корректно => ответы для 3,4 становятся неверными.")

# ---------- 1.4 ----------
print("\n=== 1.4: без H на control ===")
rows = []
for i in [1,2,3,4]:
    c = sim.run(transpile(deutsch_circuit(make_oracle(i), h_control=False), sim),
                shots=1024).result().get_counts()
    b = max(c, key=c.get)
    a_ = "constant" if b == "0" else "balanced"
    rows.append((i, b, a_, TRUTH[i], a_ == TRUTH[i]))
print(pd.DataFrame(rows, columns=["oracle","bit","answer","truth","ok"]).to_string(index=False))
print("Пояснение: без H на control кубит остаётся в |0>, суперпозиции нет,")
print("интерференция не работает => ответы некорректны.")

# ---------- 1.5 ----------
print("\n=== 1.5: deutsch_decide ===")
def deutsch_decide(oracle):
    gate = oracle.to_gate() if isinstance(oracle, QuantumCircuit) else oracle
    qc = QuantumCircuit(2, 1)
    qc.h(0); qc.x(1); qc.h(1)
    qc.append(gate, [0, 1])
    qc.h(0); qc.measure(0, 0)
    c = sim.run(transpile(qc, sim), shots=256).result().get_counts()
    return "constant" if max(c, key=c.get) == "0" else "balanced"

for i in [1,2,3,4]:
    assert deutsch_decide(make_oracle(i)) == TRUTH[i]
print("4 стандартных оракула: OK")

# Исправленная генерация 100 случайных оракулов
for _ in range(100):
    f0, f1 = np.random.randint(0, 2, 2)
    qc_o = QuantumCircuit(2)
    if f0 == 1: qc_o.cx(0, 1)       # f(0)=1 -> инвертируем target при x=0
    if f1 == 1: qc_o.x(0); qc_o.cx(0, 1); qc_o.x(0) # f(1)=1 -> инвертируем при x=1

    res = deutsch_decide(qc_o.to_gate())
    truth = "constant" if f0 == f1 else "balanced"
    assert res == truth, f"f0={f0}, f1={f1}, got {res}, truth {truth}"
print("100 случайных функций: OK (оракул вызывается ровно 1 раз)")
```

```python
# ============================================================
# ЗАДАНИЕ 2: АЛГОРИТМ ДОЙЧА-ЙОЖИ (ИСПРАВЛЕННЫЙ)
# ============================================================
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from math import log2, comb  # Используем comb для сочетаний
from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator

np.random.seed(7)
sim = AerSimulator()
N_DJ = 2

def dj_from_truth_table(f_vals):
    n = int(round(log2(len(f_vals)))); assert 2**n == len(f_vals)
    qc = QuantumCircuit(n+1)
    for x, fx in enumerate(f_vals):
        if fx:
            for i in range(n):
                if ((x >> (n-1-i)) & 1) == 0: qc.x(i)
            qc.mcx(list(range(n)), n)
            for i in range(n):
                if ((x >> (n-1-i)) & 1) == 0: qc.x(i)
    return qc

def dj_algorithm(oracle_qc, n, shots=1024):
    qc = QuantumCircuit(n+1, n)
    qc.x(n)
    for i in range(n+1): qc.h(i)
    qc.append(oracle_qc.to_gate(), range(n+1))
    for i in range(n): qc.h(i)
    qc.measure(range(n), range(n))
    return sim.run(transpile(qc, sim), shots=shots).result().get_counts()

def truth_from_oracle(oracle_qc, n):
    f = []
    for x in range(2**n):
        qc = QuantumCircuit(n+1)
        for i in range(n):
            if (x >> (n-1-i)) & 1: qc.x(i)
        qc.append(oracle_qc.to_gate(), range(n+1))
        sv = Statevector(qc)
        f.append(1 if sv.probabilities([n])[1] > .5 else 0)
    return f

print("=== 2.1: 5 случайных оракулов, n=2 ===")
rows = []
for _ in range(5):
    f_vals = list(np.random.randint(0, 2, 2**N_DJ))
    oq = dj_from_truth_table(f_vals)
    c = dj_algorithm(oq, N_DJ)
    bits = max(c, key=c.get)
    ans = "constant" if set(bits) == {"0"} else "balanced"
    true_f = truth_from_oracle(oq, N_DJ)
    truth = "constant" if all(v==true_f[0] for v in true_f) else "balanced"
    rows.append((str(true_f), bits, ans, truth, ans == truth))
print(pd.DataFrame(rows, columns=["f(x)","bits","ans","truth","ok"]).to_string(index=False))

print("\n=== 2.2: Проверка dj_from_truth_table ===")
for n_t in [2, 3, 4]:
    for kind in ["constant", "balanced"]:
        for _ in range(50):
            if kind == "constant":
                f = [np.random.randint(0,2)] * (2**n_t)
            else:
                f = [0]*(2**(n_t-1)) + [1]*(2**(n_t-1)); np.random.shuffle(f)
            oq = dj_from_truth_table(f)
            c = dj_algorithm(oq, n_t, shots=512)
            bits = max(c, key=c.get)
            ans = "constant" if set(bits) == {"0"} else "balanced"
            assert ans == kind, (n_t, kind, bits)
    print(f"n={n_t}: 50 constant + 50 balanced OK")

print("\n=== 2.3: Не constant/не balanced, n=3 ===")
def p0(f):
    n = int(round(log2(len(f))))
    s = sum((-1)**v for v in f)
    return (s / 2**n)**2
for ones in [1, 2, 3]:
    f = [1]*ones + [0]*(8-ones); np.random.shuffle(f)
    oq = dj_from_truth_table(f)
    c = dj_algorithm(oq, 3, shots=20000)
    p_sim = c.get("000", 0) / 20000
    print(f"ones={ones}: P_theory(000)={p0(f):.4f}, P_sim={p_sim:.4f}")

print("\n=== 2.4: n=2..7 ресурсы ===")
rows = []
for n_t in range(2, 8):
    f = list(np.random.randint(0, 2, 2**n_t))
    oq = dj_from_truth_table(f)
    qc = QuantumCircuit(n_t+1, n_t)
    qc.x(n_t)
    for i in range(n_t+1): qc.h(i)
    qc.append(oq.to_gate(), range(n_t+1))
    for i in range(n_t): qc.h(i)
    qc.measure(range(n_t), range(n_t))
    depth = transpile(qc, sim, optimization_level=1).depth()
    rows.append((n_t, n_t+1, depth, 2**(n_t-1)+1))
df = pd.DataFrame(rows, columns=["n","qubits","depth","class_det_worst"])
print(df.to_string(index=False))

# ИСПРАВЛЕННАЯ ФУНКЦИЯ ВЕРОЯТНОСТИ ОШИБКИ
def p_err(n, q):
    return 2 * comb(2**(n-1), q) / comb(2**n, q)

print(f"\nP_err(n=3,q=3) = {p_err(3,3):.5f}")

rows = []
for n_t in range(2, 8):
    q = 1
    while p_err(n_t, q) > 0.01: q += 1
    rows.append((n_t, q, 2**(n_t-1)+1, 1))
dfq = pd.DataFrame(rows, columns=["n","class_prob","class_det","quantum"])
print(dfq.to_string(index=False))

# Построение графика
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(dfq.n, dfq.class_det, 'o-', label="class. det.")
ax.plot(dfq.n, dfq.class_prob, 's-', label="class. prob.")
ax.plot(dfq.n, dfq.quantum, '^-', label="quantum")
ax.set_yscale("log");
ax.set_xlabel("n");
ax.set_ylabel("queries")
ax.legend()
plt.title("Сравнение числа запросов к оракулу")
plt.show()
```

```python
# ============================================================
# ЗАДАНИЕ 3: АЛГОРИТМ ШОРА (ТАБЛИЦА А)
# ============================================================
import numpy as np, pandas as pd
from math import gcd, log2, ceil
from fractions import Fraction
import sympy
from qiskit import QuantumCircuit, transpile
from qiskit.circuit.library import UnitaryGate, QFTGate # Используем QFTGate
from qiskit_aer import AerSimulator

sim = AerSimulator()
N, a = 15, 7  # Вариант 1
n = ceil(log2(N)); m = 2*n
assert gcd(a, N) == 1

def classical_order(a, N):
    r, x = 1, a % N
    while x != 1: x = (x*a) % N; r += 1
    return r

r = classical_order(a, N)
print(f"N={N}, a={a}, r={r}, phi(r)/r={sympy.totient(r)/r:.4f}")

# ---------- 3.2: Построение схемы ----------
def mod_mult_gate(b, N, n):
    dim = 2**n; M = np.zeros((dim, dim))
    for x in range(dim):
        y = (b*x) % N if x < N else x; M[y, x] = 1
    return UnitaryGate(M)

def build_shor(N, a, m, n):
    qc = QuantumCircuit(m+n, m)
    qc.x(m) # целевой кубит в |1>
    for i in range(m): qc.h(i) # H на управляющий регистр

    # Управляемые M_b
    for k in range(m):
        b = pow(a, 2**k, N)
        if b == 1: continue
        qc.append(mod_mult_gate(b, N, n).control(1), [k] + list(range(m, m+n)))

    # Обратное QFT (с использованием QFTGate)
    qc.append(QFTGate(m).inverse(), range(m))

    qc.measure(range(m), range(m))
    return qc

print("\n=== 3.2: Построение схемы ===")
qc = build_shor(N, a, m, n)
tqc = transpile(qc, sim, optimization_level=1)
print(f"Глубина схемы: {tqc.depth()}")

# ---------- 3.3: Запуск ----------
print("\n=== 3.3: Запуск схемы ===")
counts = sim.run(tqc, shots=2000).result().get_counts()
print("Топ-3 исходов:", sorted(counts.items(), key=lambda kv: -kv[1])[:3])

def process(counts, N, a, m):
    good = 0; rs = []
    for bits, c in counts.items():
        rc = Fraction(int(bits, 2), 2**m).limit_denominator(N).denominator
        if rc > 1 and pow(a, rc, N) == 1:
            good += c
            rs.append(rc)
    return good/sum(counts.values()), rs

frac, rs = process(counts, N, a, m)
print(f"Доля засчитанных: {frac:.4f} (теория: {sympy.totient(r)/r:.4f})")

# ---------- 3.4: Поиск множителей ----------
print("\n=== 3.4: Поиск множителей ===")
def factor_from_r(a, r, N):
    if r % 2: return None, "r odd"
    x = pow(a, r//2, N)
    if x == N-1: return None, "a^{r/2}=-1"
    d1, d2 = gcd(x-1, N), gcd(x+1, N)
    return (d1, d2), "ok"

res, why = factor_from_r(a, r, N)
print(f"Результат: {res} (причина: {why})")

# ---------- 3.5: Разные m ----------
print("\n=== 3.5: Разные m ===")
rows = []
for m_t in [n+1, 2*n-1, 2*n]:
    tqc_t = transpile(build_shor(N, a, m_t, n), sim, optimization_level=1)
    c_t = sim.run(tqc_t, shots=1000).result().get_counts()
    fr, _ = process(c_t, N, a, m_t)
    rows.append((m_t, fr, sympy.totient(r)/r))
print(pd.DataFrame(rows, columns=["m","p_hat","phi(r)/r"]).to_string(index=False))
```

```python
# ============================================================
# ЗАДАНИЕ 4: СТАТИСТИКА ПОВТОРОВ
# ============================================================
import numpy as np, pandas as pd
from math import gcd, log2, ceil
from fractions import Fraction
import sympy
from qiskit import QuantumCircuit, transpile
from qiskit.circuit.library import UnitaryGate, QFTGate
from qiskit_aer import AerSimulator

sim = AerSimulator()
N, a = 15, 7
n = ceil(log2(N)); m = 2*n

def classical_order(a, N):
    r, x = 1, a % N
    while x != 1: x = (x*a) % N; r += 1
    return r

r = classical_order(a, N)
# Явно приводим к float, чтобы избежать ошибки с sympy
p_theory = float(sympy.totient(r)) / r

def mod_mult_gate(b, N, n):
    dim = 2**n; M = np.zeros((dim, dim))
    for x in range(dim):
        y = (b*x) % N if x < N else x; M[y, x] = 1
    return UnitaryGate(M)

def build(N, a, m, n):
    qc = QuantumCircuit(m+n, m); qc.x(m)
    for i in range(m): qc.h(i)
    for k in range(m):
        b = pow(a, 2**k, N)
        if b == 1: continue
        qc.append(mod_mult_gate(b, N, n).control(1), [k] + list(range(m, m+n)))
    # Используем QFTGate вместо QFT
    qc.append(QFTGate(m).inverse(), range(m))
    qc.measure(range(m), range(m))
    return qc

def accepted_fraction(counts, N, a, m):
    good = 0
    for bits, c in counts.items():
        rc = Fraction(int(bits, 2), 2**m).limit_denominator(N).denominator
        if rc > 1 and pow(a, rc, N) == 1: good += c
    return good/sum(counts.values())

tqc = transpile(build(N, a, m, n), sim, optimization_level=1)

print("=== 4.1: 5 прогонов по 1024 shots ===")
ps = []
for seed in range(5):
    c = AerSimulator(seed_simulator=seed).run(tqc, shots=1024).result().get_counts()
    p = accepted_fraction(c, N, a, m); ps.append(p)

print(f"mean={np.mean(ps):.4f}, std={np.std(ps):.4f}, phi(r)/r={p_theory:.4f}")

print("\n=== 4.2: Оценка числа повторов ===")
# Теперь p_theory - обычный float, np.log отработает без ошибок
T = int(np.ceil(np.log(0.01) / np.log(1 - p_theory)))
print(f"Теоретическое T = {T} (для 99% успеха)")
```

```python
# ============================================================
# ЗАДАНИЕ 5: ЭМУЛЯЦИЯ (ТАБЛИЦА Б)
# ============================================================
import numpy as np, pandas as pd, psutil
from math import gcd, log2, ceil, pi, sin
from fractions import Fraction
from collections import Counter
import sympy

np.random.seed(11)
N, a1, a2 = 67721, 3, 2  # Вариант 16
n = ceil(log2(N)); m = 2*n

print("=== 5.1: Память ===")
mem = 16 * 2**(3*n)
ram = psutil.virtual_memory().total
print(f"Нужно кубитов: {3*n}, память: {mem/2**30:.2e} GiB, RAM: {ram/2**30:.1f} GiB")

def P_y(y, theta, m):
    delta = 2**m * theta - y
    if abs(delta) < 1e-12: return 1.0
    return (sin(pi*delta) / (2**m * sin(pi*delta / 2**m)))**2

def classical_order(a, N):
    r, x = 1, a % N
    while x != 1: x = (x*a) % N; r += 1
    return r

def process_ys(ys, N, a, m):
    good = 0
    for y in ys:
        rc = Fraction(int(y), 2**m).limit_denominator(N).denominator
        if rc > 1 and pow(a, rc, N) == 1: good += 1
    return good/len(ys)

def emulate(a_val, N, m, n_samples=2000, window=300):
    r = classical_order(a_val, N)
    ys = []
    for _ in range(n_samples):
        k = np.random.randint(0, r)
        th = k/r; c = 2**m*th
        cand = np.arange(max(0, int(c)-window), min(2**m-1, int(c)+window)+1)
        w = np.array([P_y(int(y), th, m) for y in cand]); w /= w.sum()
        ys.append(np.random.choice(cand, p=w))
    return r, ys

print("\n=== 5.3: Эмуляция для a1 ===")
r1, ys1 = emulate(a1, N, m)
frac1 = process_ys(ys1, N, a1, m)
print(f"a1={a1}: r={r1}, acc={frac1:.4f}, phi(r)/r={sympy.totient(r1)/r1:.4f}")

print("\n=== 5.6: Минимальное m ===")
target = 0.9 * sympy.totient(r1)/r1
for m_t in range(n, 2*n+1):
    _, ys = emulate(a1, N, m_t, n_samples=1000)
    fr = process_ys(ys, N, a1, m_t)
    print(f"m={m_t}: acc={fr:.4f} (target={target:.4f})")
    if fr >= target:
        print(f"Минимальное m = {m_t}"); break
```

```python
# ============================================================
# ЗАДАНИЕ 6: shor_factor(N, bases)
# ============================================================
import numpy as np, sympy
from math import gcd, log2, ceil, pi, sin
from fractions import Fraction
from collections import Counter
from qiskit import QuantumCircuit, transpile
from qiskit.circuit.library import QFTGate, UnitaryGate
from qiskit_aer import AerSimulator

sim = AerSimulator()

# ---------- Вспомогательные функции ----------
def classical_order(a, N):
    r, x = 1, a % N
    while x != 1: x = (x*a) % N; r += 1
    return r

def factor_from_r(a, r, N):
    """Пытается найти нетривиальные множители по порядку r."""
    if r % 2: return None
    x = pow(a, r//2, N)
    if x in (1, N-1): return None
    d1, d2 = gcd(x-1, N), gcd(x+1, N)
    if 1 < d1 < N: return (d1, N//d1)
    if 1 < d2 < N: return (d2, N//d2)
    return None

def mod_mult_gate(b, N, n):
    dim = 2**n; M = np.zeros((dim, dim))
    for x in range(dim):
        y = (b*x) % N if x < N else x; M[y, x] = 1
    return UnitaryGate(M)

def P_y(y, theta, m):
    delta = 2**m * theta - y
    if abs(delta) < 1e-12: return 1.0
    return (sin(pi*delta) / (2**m * sin(pi*delta / 2**m)))**2

# ---------- Полная схема (Таблица А) ----------
def run_qpe(N, a, n_samples=1024):
    n = ceil(log2(N)); m = 2*n
    qc = QuantumCircuit(m+n, m); qc.x(m)
    for i in range(m): qc.h(i)
    for k in range(m):
        b = pow(a, 2**k, N)
        if b == 1: continue
        qc.append(mod_mult_gate(b, N, n).control(1), [k] + list(range(m, m+n)))
    qc.append(QFTGate(m).inverse(), range(m))
    qc.measure(range(m), range(m))

    tqc = transpile(qc, sim, optimization_level=1)
    counts = sim.run(tqc, shots=n_samples).result().get_counts()

    rs = []
    for bits, cnt in counts.items():
        rc = Fraction(int(bits, 2), 2**m).limit_denominator(N).denominator
        if rc > 1 and pow(a, rc, N) == 1:
            rs.extend([rc] * cnt)
    return rs

# ---------- Эмуляция (Таблица Б) ----------
def run_emulation(N, a, n_samples=1024, window=300):
    n = ceil(log2(N)); m = 2*n
    r = classical_order(a, N)
    ys = []
    for _ in range(n_samples):
        k = np.random.randint(0, r)
        th = k/r
        center = 2**m * th
        cand = np.arange(max(0, int(center)-window), min(2**m-1, int(center)+window)+1)
        w = np.array([P_y(int(y), th, m) for y in cand])
        w /= w.sum()
        ys.append(np.random.choice(cand, p=w))

    rs = []
    for y in ys:
        rc = Fraction(int(y), 2**m).limit_denominator(N).denominator
        if rc > 1 and pow(a, rc, N) == 1:
            rs.append(rc)
    return rs

# ---------- Главная функция ----------
def shor_factor(N, bases, use_emulation=None):
    """
    Возвращает (p, q), основание, номер попытки.
    Автоматически выбирает между симуляцией и эмуляцией.
    """
    if sympy.isprime(N):
        raise ValueError(f"N={N} — простое число, множителей нет!")

    if use_emulation is None:
        # Если N слишком велико (нужно > 20 кубитов), симулятор упадет по памяти
        use_emulation = (ceil(log2(N)) > 10)

    for attempt, a in enumerate(bases, 1):
        if gcd(a, N) != 1:
            continue

        # Получаем порядок r
        if use_emulation:
            rs = run_emulation(N, a)
        else:
            rs = run_qpe(N, a)

        if not rs:
            continue

        # Берем самый частый порядок
        r_mode = Counter(rs).most_common(1)[0][0]

        # Пытаемся найти множители
        fac = factor_from_r(a, r_mode, N)
        if fac:
            return fac, a, attempt

    return None, None, len(bases)

# ============================================================
# ТЕСТИРОВАНИЕ И ОТВЕТЫ НА ВОПРОСЫ
# ============================================================
if __name__ == "__main__":
    print("=== 6: Тестирование shor_factor ===")

    # Тест 1: Малое N (Таблица А, полная схема)
    N_A = 15
    fac_A, base_A, att_A = shor_factor(N_A, bases=[7, 11])
    print(f"N={N_A}: множители={fac_A}, база={base_A}, попыток={att_A}")
    assert fac_A[0] * fac_A[1] == N_A

    # Тест 2: Большое N (Таблица Б, эмуляция)
    N_B = 67721
    fac_B, base_B, att_B = shor_factor(N_B, bases=[3, 2])
    print(f"N={N_B}: множители={fac_B}, база={base_B}, попыток={att_B}")
    assert fac_B[0] * fac_B[1] == N_B

    print("\n=== Ответы на вопросы ===")
    print("1. Если N простое (например, 1009), алгоритм выдаст ошибку (ValueError).")
    print("   Для простого N все основания a взаимно просты с N, но порядок r всегда")
    print("   будет делить N-1. При этом gcd(a^(r/2) - 1, N) всегда будет равен 1 или N.")
    print("   Поэтому нетривиальный множитель получить невозможно.")
    print("2. Нечётный порядок r не даёт множителей, потому что в формуле Шора")
    print("   используется a^(r/2). Если r нечётно, r/2 не является целым числом,")
    print("   и мы не можем вычислить этот корень в кольце вычетов.")
    print("3. Для N длиной 2048 бит потребуется 3 * 2048 = 6144 логических кубита.")
    print("   (Формула 3*ceil(log2(N)), где для 2048 бит ceil(log2(N)) = 2048).")
```
